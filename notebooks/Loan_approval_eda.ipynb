```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Data\loan_approval_dataset.csv")
df.columns = df.columns.str.strip()
```

```python
df["loan_status"].value_counts().plot(
    kind="bar"
)

plt.title("Loan Approval Distribution")
plt.xlabel("Loan Status")
plt.ylabel("Count")
plt.show()
```

```python
df["education"].value_counts().plot(
    kind="bar"
)

plt.title("Education Distribution")
plt.show()
```

```python
df["self_employed"].value_counts().plot(
    kind="bar"
)

plt.title("Self Employed")
plt.show()
```

```python
plt.hist(
    df["income_annum"],
    bins=20
)

plt.title("Income Distribution")
plt.xlabel("Annual Income")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(
    df["loan_amount"],
    bins=20
)

plt.title("Loan Amount Distribution")
plt.show()
```

```python
plt.hist(
    df["cibil_score"],
    bins=20
)

plt.title("CIBIL Score")
plt.show()
```

```python
plt.scatter(
    df["income_annum"],
    df["loan_amount"]
)

plt.title("Income vs Loan Amount")
plt.xlabel("Income")
plt.ylabel("Loan Amount")
plt.show()
```

```python
plt.scatter(
    df["cibil_score"],
    df["loan_amount"]
)

plt.title("CIBIL Score vs Loan Amount")
plt.show()
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(10,8))

plt.imshow(corr)

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.colorbar()

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.boxplot(
    df["income_annum"]
)

plt.title("Income Outliers")

plt.show()
```

```python
plt.boxplot(
    df["loan_amount"]
)

plt.title("Loan Amount Outliers")

plt.show()
```

```python
approved = df[df["loan_status"]==" Approved"]["cibil_score"]
rejected = df[df["loan_status"]==" Rejected"]["cibil_score"]

plt.hist(
    approved,
    alpha=0.6,
    label="Approved"
)

plt.hist(
    rejected,
    alpha=0.6,
    label="Rejected"
)

plt.legend()

plt.title("CIBIL Score by Loan Status")

plt.show()
```
